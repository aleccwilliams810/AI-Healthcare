```python
query = """
SELECT
    p.SUBJECT_ID,
    p.GENDER,
    p.DOB,
    CASE
        WHEN t.TRANSFER_FLAG IS NOT NULL THEN 1
        ELSE 0
    END AS TRANSFER_FLAG
FROM
    `physionet-data.mimiciii_clinical.patients` p
LEFT JOIN (
    SELECT
        SUBJECT_ID,
        1 AS TRANSFER_FLAG
    FROM
        `physionet-data.mimiciii_clinical.transfers`
    WHERE
        eventtype = 'transfer'
    GROUP BY
        SUBJECT_ID
) t
ON
    p.SUBJECT_ID = t.SUBJECT_ID
"""

# Run query and load into a DataFrame
query_job = client.query(query)
df = query_job.result().to_dataframe()
df.to_csv('patients.csv', index=False)
```
